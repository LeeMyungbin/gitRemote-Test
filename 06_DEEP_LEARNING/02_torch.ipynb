## 연산

```python
a = torch.tensor([[1,2,3],[4,5,6]], dtype=torch.float)
b = torch.tensor([[7,8,9],[10,11,12]], dtype=torch.float)
c = torch.tensor([[1,2], [3,4],[5,6]], dtype=torch.float)

print(a.shape, b.shape, c.shape)
```

```python
# 원소별 연산
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
```

```python
# broadcasting 연산
print(a + 1)
print(a + torch.tensor([10,20,30]))
# 마지막 축부터 확인하는데 (2,3) 과 (1,2)는 3과 2로 같지 않아 연산 불가
# print(a + torch.tensor([10,20]))
```

```python
# 내적 (모든 원소간의 곱을 더하기)
# - 1차원 : torch.dot(), torch.matmul(), @
# - 2차원 : torch.matmul(), @
m = torch.tensor([1,2,3])
n = torch.tensor([4,5,6])
print(m.dot(n))
print(torch.dot(m,n))
print(torch.matmul(m,n)) 
print(m @ n)
```

```python
# 2차원 이상의 내적
print(a.matmul(c))   # (2,3) @ (3,2) = > (2,2)
print(a @ c)
# print(torch.dot(a,c)) -> 2차원에서 사용 불가
```

## 형태 변경

```python
t = torch.arange(12)
print(t)

t_ = t.reshape(3,4)
t_ = t.reshape(3,-1)
t_ = t.reshape(-1,4)
t_ = t.reshape(2,2,3)

print(t_, t_.size())
```

```python
# 평탄화
print(t.flatten())
print(t.reshape(-1))
```

## indexing | slicing

```python
x =torch.arange(1,13).reshape(3,2,2)
x
```

```python
# 인덱싱
print(x[0])
print(x[1,0])
print(x[2,1])
print(x[-1,-1])
print(x[-1,0,-1])
```

```python
# 슬라이싱
x_ = x[:,0]    # 인덱싱 결과는 2차원
x_ = x[:,:1]   # 슬라이싱 결과는 3차원

print(x_, x_.shape)
```

```python
# fancy indexing
index = [1,2]
x_ = x[index]

print(x_, x_.shape)
```

## 유용한 메소드

```python
# split
x = torch.zeros(10,4)
print(x)

x_splits = torch.split(x,4)

for x_ in x_splits:
    print(x_)
```

```python
# expand
k = torch.tensor([[[1,2]], [[3,4]]])
print(k, k.shape)
k_exp = k.expand(2,3,2)
print(k_exp, k_exp.shape)
```

```python
# 원본 배열 k의 view 형태로 k_exp가 생성 되어 k_exp가 k를 참조하고 있다
k_exp[0,0,0] = 100
print(k)
print(k_exp)
```

```python
# repeat
k = torch.tensor([[[1, 2]], [[3, 4]]])
print(k, k.shape)

# 차원별 반복 횟수
k_rep = k.repeat(1, 3, 1)
print(k_rep, k_rep.shape)

k_rep[0, 0, 0] = 100
print(k)
print(k_rep)
```

```python
# masked_fill: mask 조건식이 true 경우만 값 변경
x = torch.arange(9).reshape(3,-1)
print(x)

mask = x >4
print(mask)

x_ = x.masked_fill(mask, value=0)
print(x_)
```

## 모델 학습에서 tensor 다차원 배열 연산

```python
# 단중 선형 연산
X = torch.randn(4,5)  # 연력 값: 4개의 sample, 5개의 feature
print(f'{X=}')
W = torch.randn()     # 가중치 (3개의 노드)
print(f'{W=}')
b = torch.randn(1,3)  # 노드마다 편향(bias)
print(f'b=')

# Linear 총 계산
z = torch.matmul(X, W) +b

z
```
